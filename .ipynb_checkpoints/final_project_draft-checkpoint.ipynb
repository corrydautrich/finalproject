```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("data/Olympic_Swimming_Results_1912to2020.csv")
df.head()
```

I loaded the Olympic swimming dataset from 1912–2020. 
The dataset includes columns such as Athlete, Team, Gender, Stroke, Distance, Year, Rank, and Results. 

Key characteristics observed:
- There are some missing values in the Results column, which we will handle by removing rows with missing target values.
- The Results column contains times as strings, which we convert to seconds for regression.
- The dataset includes both categorical (Stroke, Gender) and numeric (Distance, Year) features.

```python
print("Shape:", df.shape)
print("\nData Types:\n", df.dtypes)
print("\nSummary Stats:\n", df.describe())
```

```python
def time_to_seconds(t):
    try:
        # if t is already numeric, return it
        if isinstance(t, (float, int)):
            return t
        # split by colon
        parts = t.split(":")
        if len(parts) == 1:
            return float(parts[0])
        elif len(parts) == 2:
            return int(parts[0])*60 + float(parts[1])
        elif len(parts) == 3:
            return int(parts[0])*3600 + int(parts[1])*60 + float(parts[2])
    except:
        return np.nan  # if conversion fails
df["Results_float"] = df["Results"].apply(time_to_seconds)
df["Medal"] = df["Rank"].map(lambda x: "Gold" if x==1 else ("Silver" if x==2 else ("Bronze" if x==3 else "None")))
target = "Results_float"
df[target].hist(bins=30)
plt.title("Distribution of Swim Times (seconds)")
plt.xlabel("Time (s)")
plt.ylabel("Count")
plt.show()
df["Medal"].value_counts().plot(kind='bar')
plt.title("Medal Distribution")
plt.show()
```

I visualized the distribution of swim times and medals to understand the target variable and class distribution:

- Histogram of Results (in seconds) to see the distribution of swim times.
- Bar chart of Medal counts to observe class balance for top 3 medals.

```python
df["Distance_m"] = df["Distance (in meters)"].str.replace(r"\D", "", regex=True).astype(float)

# One-hot encode categorical variables
df = pd.get_dummies(df, columns=["Stroke", "Gender"], drop_first=True)

# Interaction feature example (safe in case Gender_Male doesn't exist)
df["IsRelayAndMale"] = df["Relay?"] * df.get("Gender_Male", 0)

# Normalize year
df["Year_Since_2000"] = df["Year"] - 2000
```

I created 3 engineered features based on the dataset:

1. **Results_float**: Converted race results from strings to seconds to create a numeric target for regression.
   - Helps regression models learn better since numeric values are needed.
2. **Distance_m**: Converted "Distance (in meters)" to numeric float values.
   - Captures the actual race distance in meters for use as a numeric feature.
3. **IsRelayAndMale**: Interaction feature between Relay status and Gender (Male).
   - This captures the potential difference in performance patterns for male relay participants.
4. **Year_Since_2000**: Normalized the Year column.
   - Helps models learn trends over time while avoiding large numbers.

```python
# Regression
X_reg = df.drop(columns=["Results_float", "Medal", "Athlete", "Team", "Location", "Results", "Rank", "Distance (in meters)"])
y_reg = df["Results_float"]

# Classification
X_clf = X_reg.copy()
y_clf = df["Medal"]
# Fill numeric missing values with mean
numeric_cols = X_reg.select_dtypes(include=["float64", "int64"]).columns
X_reg[numeric_cols] = X_reg[numeric_cols].fillna(X_reg[numeric_cols].mean())
X_clf[numeric_cols] = X_clf[numeric_cols].fillna(X_clf[numeric_cols].mean())
```

- Kept only rows where the target `Results_float` is not missing.
- Split dataset into training (80%) and test (20%) sets.
- Scaled numeric features using StandardScaler to normalize feature ranges.
- One-hot encoded categorical features (Stroke, Gender) to convert to numeric representation for modeling.

```python
from sklearn.model_selection import train_test_split

# Regression
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg, y_reg, test_size=0.2, random_state=42
)

# Classification (stratify to keep class balance)
X_train_clf, X_test_clf, y_train_clf, y_test_clf = train_test_split(
    X_clf, y_clf, test_size=0.2, random_state=42, stratify=y_clf
)
```

```python
from sklearn.preprocessing import StandardScaler

scaler_reg = StandardScaler()
X_train_reg_scaled = scaler_reg.fit_transform(X_train_reg)
X_test_reg_scaled = scaler_reg.transform(X_test_reg)

scaler_clf = StandardScaler()
X_train_clf_scaled = scaler_clf.fit_transform(X_train_clf)
X_test_clf_scaled = scaler_clf.transform(X_test_clf)
```

- Linear Regression is a baseline model to predict swim times (continuous target).
- Features are scaled numeric features and one-hot encoded categorical features.

```python
# Keep only rows where target is not NaN
reg_df = df.dropna(subset=["Results_float"])

# Prepare features and target again
X_reg = reg_df.drop(columns=["Results_float", "Medal", "Athlete", "Team", "Location", "Results", "Rank", "Distance (in meters)"])
y_reg = reg_df["Results_float"]

# Split train/test
from sklearn.model_selection import train_test_split
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg, y_reg, test_size=0.2, random_state=42
)

# Scale numeric features
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train_reg_scaled = scaler.fit_transform(X_train_reg)
X_test_reg_scaled = scaler.transform(X_test_reg)

# Train model
from sklearn.linear_model import LinearRegression
reg_model = LinearRegression()
reg_model.fit(X_train_reg_scaled, y_train_reg)

# Predict and evaluate
y_pred_reg = reg_model.predict(X_test_reg_scaled)
from sklearn.metrics import mean_squared_error, r2_score
rmse = mean_squared_error(y_test_reg, y_pred_reg) ** 0.5
print("Regression RMSE:", rmse)
print("Regression R2:", r2_score(y_test_reg, y_pred_reg))
```

The model provides a baseline for swim time prediction. Observed errors suggest feature scaling and interactions may improve performance. A scatter plot of predicted vs actual results can visualize performance.

```python
# Coefficients of features
feature_names = X_reg.columns
coef = reg_model.coef_

importance_df = pd.DataFrame({
    "Feature": feature_names,
    "Coefficient": coef
}).sort_values(by="Coefficient", key=abs, ascending=False)

print(importance_df)
```

```python
import matplotlib.pyplot as plt

residuals = y_test_reg - y_pred_reg

plt.scatter(y_pred_reg, residuals)
plt.axhline(0, color='red', linestyle='--')
plt.xlabel("Predicted Time (s)")
plt.ylabel("Residuals")
plt.title("Residual Plot")
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix

X_clf = df.drop(columns=["Results_float", "Medal", "Athlete", "Team", "Location", "Results", "Rank", "Distance (in meters)"])
y_clf = df["Medal"]

X_train_clf, X_test_clf, y_train_clf, y_test_clf = train_test_split(
    X_clf, y_clf, test_size=0.2, random_state=42
)

clf_model = RandomForestClassifier(n_estimators=100, random_state=42)
clf_model.fit(X_train_clf, y_train_clf)
y_pred_clf = clf_model.predict(X_test_clf)

print(classification_report(y_test_clf, y_pred_clf))
```

- I chose Random Forest to capture non-linear relationships between features and swim times.
- Features and preprocessing are the same as for Linear Regression.

```python
plt.scatter(y_test_reg, y_pred_reg, alpha=0.5)
plt.plot([y_test_reg.min(), y_test_reg.max()], [y_test_reg.min(), y_test_reg.max()], 'r--')
plt.xlabel("Actual Time (s)")
plt.ylabel("Predicted Time (s)")
plt.title("Predicted vs Actual Swim Times")
plt.show()
```

```python
feature_importance = pd.DataFrame({
    "Feature": X_reg.columns,
    "Coefficient": reg_model.coef_
}).sort_values(by="Coefficient", key=abs, ascending=False)

print(feature_importance)
```
